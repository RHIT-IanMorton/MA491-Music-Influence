```python
# Find largest scores with mariah carrey, most similar artists and their scores.
check_artist = "262255"     # mariah carrey
check_vec = get_artist_vector(check_artist)

artist_ids = get_all_ids(FEATURE_FILE)
artist_ids.remove(check_artist)
vectors = get_artist_vectors(artist_ids)


max = 0
closest = ""
scores = {}
for id in artist_ids:
    s = score(check_vec, vectors.loc[id], method="minkowski")
    scores[id] = s
    if s > max:
        max = s
        closest = id

print(f"max: {max} with {closest}")

sorted_data = dict(sorted(scores.items(), key=lambda item: item[1], reverse=True))
print("top scores in order: ",sorted_data)
```

```python
metallica = get_artist_vector("446509", ARTIST_FILE)
mariah = get_artist_vector("262255", ARTIST_FILE)
metallica_std = get_artist_vector("446509", FEATURE_FILE)
mariah_std = get_artist_vector("262255", FEATURE_FILE)
halen_std = get_artist_vector("260206", FEATURE_FILE)

score_norm = np.dot(metallica / np.linalg.norm(metallica),  mariah / np.linalg.norm(mariah))
score_std = np.dot(metallica_std / np.linalg.norm(metallica_std),  mariah_std / np.linalg.norm(mariah_std))
score_std_dist = score(metallica_std, mariah_std)

print("norms: ",np.linalg.norm(metallica_std), np.linalg.norm(mariah_std), np.linalg.norm(halen_std))
print("normal score:", score_norm, "\nstandardized score: ", score_std, "\nstandardized distance score: ", score_std_dist)
```
